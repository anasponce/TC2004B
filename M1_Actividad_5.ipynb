**1. How many men and women (*sex* feature) are represented in this dataset?**

```python
male = data.loc[data["sex"] == "Male"]
female = data.loc[data["sex"] == "Female"]

print(len(male), "men")
print(len(female), "women")
```

**2. What is the average age (*age* feature) of women?**

```python
fem_age = data.age[data["sex"] == "Female"]
print(round(fem_age.mean()), "years")
```

**3. What is the percentage of German citizens (*native-country* feature)?**

```python
german = data.loc[data["native-country"] == "Germany"]
print(round((len(german) / len(data))*100, 1), "%")
```

4-5. __What are the mean and standard deviation of age for those who earn more than 50K per year (*salary* feature) and those who earn less than 50K per year?__

```python
more50k = data.age[data["salary"] == ">50K"]
less50k = data.age[data["salary"] == "<=50K"]

print("mean of age for those who earn more than 50K per year: ", round(more50k.mean()), "years")
print("mean of age for those who earn less than 50K per year: ", round(less50k.mean()), "years")

print("standard deviation of age for those who earn more than 50K per year: ", round(more50k.std()), "years")
print("standard deviation of age for those who earn less than 50K per year: ", round(less50k.std()), "years")
```

**6. Is it true that people who earn more than 50K have at least high school education? (*education – Bachelors, Prof-school, Assoc-acdm, Assoc-voc, Masters* or *Doctorate* feature)**

```python
valid_education = ["Bachelors", "Prof-school", "Assoc-acdm", "Assoc-voc", "Masters", "Doctorate"]
count_valid = 0
count_invalid = 0

for i in data.education[data["salary"] == ">50K"]:
    if i in valid_education:
        count_valid += 1
    else:
        count_invalid += 1

if count_invalid == 0:
    print("It is true")
else:
    print("It is false")

print(count_valid, "people who earn more than 50K have at least high school education")
print(count_invalid, "people who earn more than 50K do not have at least high school education")
```

**7. Display age statistics for each race (*race* feature) and each gender (*sex* feature). Use *groupby()* and *describe()*. Find the maximum age of men of *Amer-Indian-Eskimo* race.**

```python
age_stats = data.groupby(["race", "sex"])["age"].describe()
print(age_stats)
print("The maximum age of Amer-Indian-Eskimo men is 82 years")
```

**8. What is the maximum number of hours a person works per week (*hours-per-week* feature)? How many people work such a number of hours, and what is the percentage of those who earn a lot (>50K) among them?**

```python
maxhrs = data["hours-per-week"].max()
print("Maximum hours of work per week: ", maxhrs, "hrs")

people_maxhrs = data.loc[data["hours-per-week"] == 99]
print("People who work 99 hrs per week: ", len(people_maxhrs))

people_maxhrs_50k = people_maxhrs.loc[people_maxhrs["salary"] == ">50K"]
print("Percentage of those who work 99 hrs per week and earn more than 50K annually: ", round((len(people_maxhrs_50k) / len(people_maxhrs))*100), "%")
```

**9. Find the average time of work (*hours-per-week*) for those who earn a little and a lot (*salary*) for each country (*native-country*). What will these be for Japan?**

```python
workstats = data.groupby(["native-country", "salary"])["hours-per-week"].mean()
japan = workstats.loc["Japan"]
print("Average time of work per salary for each country:", workstats)
print("Japan's average time of work per salary:", japan)
```

```python
data['hours-range'] = pd.cut(data['hours-per-week'], bins = [0, 30, 40, 50, 60, 100], labels = ['0-30', '31-40', '41-50', '51-60', '61+'])
conteo = data.groupby(['hours-range', 'salary']).size().unstack(fill_value=0)
conteo.plot(kind = 'bar', stacked = True, figsize=(8, 6), color=['pink', 'skyblue'])

plt.title('Distribución de Salarios según Horas de Trabajo a la Semana')
plt.xlabel('Rango de Horas de Trabajo a la Semana')
plt.xticks(rotation = 0)
plt.ylabel('Número de Personas')
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.legend(title = 'Salario', labels = ['<=50K', '>50K'])
plt.show()
```
